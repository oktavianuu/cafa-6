```python
!pip install biopython

import pandas as pd
import numpy as np
from Bio import SeqIO
from collections import Counter
import os
import time
```

```python
base_path = ""

print("Listing files in input directory:")
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
        if "train_terms.tsv" in filename:
            base_path = dirname 

# DEFINING PATHS
DATA_DIR = base_path 
TEST_FASTA_PATH = os.path.join(os.path.dirname(base_path), "Test", "testsuperset.fasta")

OUTPUT_DIR = "./" 
NUM_PREDICTIONS = 10

print(f"\nTraining data dir: {DATA_DIR}")
print(f"Test fasta path: {TEST_FASTA_PATH}")
print(f"Output dir: {OUTPUT_DIR}")
```

```python
def get_terms_frequencies(file_path):
    print("1. Loading training terms...")
    train_terms = pd.read_csv(
        file_path,
        sep='\t',
        usecols=['EntryID', 'term'],
        header=0 
    )
    all_terms = train_terms['term'].tolist()
    term_counts = Counter(all_terms)
    most_common = term_counts.most_common(NUM_PREDICTIONS)
    
    term_freq_df = pd.DataFrame(most_common, columns=['go_term', 'frequency'])
    total_annotations = term_freq_df['frequency'].sum()
    term_freq_df['score'] = term_freq_df['frequency'] / total_annotations
    
    print(f" -> Total unique GO terms: {len(term_counts):,}")
    return term_freq_df.drop(columns=['frequency'])

def get_test_proteins(file_path):
    print("2. Loading test proteins...")
    test_proteins = []
    for record in SeqIO.parse(file_path, "fasta"):
        parts = record.id.split('|')
        protein_id = parts[1] if len(parts) > 1 else record.id
        test_proteins.append(protein_id)
    print(f" -> Total test proteins: {len(test_proteins):,}")
    return test_proteins

def generate_naive_submission(test_proteins, term_scores):
    print("3. Generating naive submission (Optimized)...")
    
    # Vectorized approach
    multi_index = pd.MultiIndex.from_product(
        [test_proteins, term_scores['go_term']], 
        names=['EntryID', 'go_term']
    )
    submission_df = multi_index.to_frame(index=False)
    submission_df = submission_df.merge(term_scores, on='go_term', how='left')
    
    # Format
    submission_df = submission_df[['EntryID', 'go_term', 'score']]
    submission_df['score'] = submission_df['score'].apply(lambda x: f"{x:.6f}")
    
    # Output File
    timestamp = time.strftime("%Y%m%d-%H%M%S")
    output_file = os.path.join(OUTPUT_DIR, f"submission_{timestamp}.tsv")

    submission_df.to_csv(output_file, sep='\t', index=False, header=False)
    print(f"4. Success! Saved to: {output_file}")
    
    return submission_df
```

```python
train_terms_file = os.path.join(DATA_DIR, "train_terms.tsv") 


if os.path.exists(train_terms_file):
    naive_scores_df = get_terms_frequencies(train_terms_file)
    print(naive_scores_df.head())
    
    if os.path.exists(TEST_FASTA_PATH):
        test_ids = get_test_proteins(TEST_FASTA_PATH)
        
        # 3. Generate Submission
        submission = generate_naive_submission(test_ids, naive_scores_df)
        
        submission.to_csv("submission.tsv", sep='\t', index=False, header=False)
        print("Also saved as 'submission.tsv' in working directory.")
        
    else:
        print(f"ERROR: Could not find Test Fasta at {TEST_FASTA_PATH}")
else:
    print(f"ERROR: Could not find Train Terms at {train_terms_file}")
```

## Experiment 2: Amino Acid Composition Features

```python
def calculate_aac_features(fasta_path):
    """
    Read fasta file and calculates freq of 20 standard aa (A, C, D, E, F, G, H, I, K, L, M, N, P, Q, R, S, 
    T, V, W, Y) for every protein
    """
    print("Calculating Amino Acid Composition features")
    protein_features = {}
    amino_acids = 'ACDEFGHIKLMNPQRSTVWY'

    # parse fasta file
    for record in SeqIO.parse(fasta_path, "fasta"):
        # extract protein id
        parts = record.id.split('|')
        protein_id = parts[1] if len(parts) > 1 else record.id

        sequence = str(record.seq).upper().replace('*', '').replace('X', '') #Clean seq
        seq_length = len(sequence)

        if seq_length == 0:
            # skip empty sequence
            continue

        # count frequencies
        counts = Counter(sequence)

        # store the freq
        features = {}
        for aa in amino_acids:
            # Store normalized frequency (count / length)
            features[aa] = counts[aa] / seq_length

        protein_features[protein_id] = features

    # convert to DataFrame
    aac_df = pd.DataFrame.from_dict(protein_features, orient='index')
    aac_df.index.name = 'EntryID'

    print(f" -> AAC features calculated for {len(aac_df):,} proteins.")
    return aac_df
```

```python
aac_features_df = calculate_aac_features(TEST_FASTA_PATH)

print("\nAAC Features Head:")
print(aac_features_df.head())
print(f"\nDataFrame shape: {aac_features_df.shape}")
```

```python
from sklearn.model_selection import train_test_split

top_10_terms = naive_scores_df['go_term'].tolist()

TRAIN_TERMS_PATH = os.path.join(DATA_DIR, "train_terms.tsv")
train_labels = pd.read_csv(
    TRAIN_TERMS_PATH,
    sep='\t',
    usecols=['EntryID', 'term'],
    header=0
)

train_labels_filtered = train_labels[train_labels['term'].isin(top_10_terms)]

Y_train_all = train_labels_filtered.pivot_table(
    index='EntryID', 
    columns='term', 
    values='term', 
    aggfunc='size', 
    fill_value=0
).apply(lambda x: x > 0).astype(int) # True/False to 1/0

print("Target variable (Y) prepared:")
print(Y_train_all.head())
print(f"Shape of Y_train: {Y_train_all.shape}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.multioutput import MultiOutputClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
```

```python
X_train_merged = aac_features_df.loc[Y_train_all.index]
Y_train = Y_train_all

print(f"X (Features) shape after merge: {X_train_merged.shape}")
print(f"Y (Labels) shape after merge: {Y_train.shape}")
print("-" * 30)
```

```python
# Preprocessing (Imputation and Scaling)
# Since we are using frequencies, we don't expect NaNs,

imputer = SimpleImputer(strategy='constant', fill_value=0)
X_train_imputed = imputer.fit_transform(X_train_merged)

# Scale the features 
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_imputed)
```

```python
# Model Definition and Training
print("Training Multi-Label Logistic Regression...")

# We use MultiOutputClassifier because we have 10 independent GO terms to predict.
# This wrapper trains 10 separate Logistic Regression models, one for each term.
classifier = MultiOutputClassifier(LogisticRegression(
    solver='liblinear', # Good for small datasets/multiclass
    random_state=42, 
    max_iter=100
), n_jobs=-1) # Use all available cores for faster training

start_time = time.time()
classifier.fit(X_train_scaled, Y_train)
end_time = time.time()

print(f"Training completed in {end_time - start_time:.2f} seconds.")

# Store the fitted scaler and imputer for use on the test set later
# (This is a good habit for machine learning reproducibility!)
fitted_scaler = scaler
fitted_imputer = imputer
```

```python
# Prepare Test Features (X_test)
print("6. Preprocessing test features...")
# Apply the SAME preprocessing steps (imputer and scaler) fitted on the training data.
X_test_imputed = fitted_imputer.transform(aac_features_df)
X_test_scaled = fitted_scaler.transform(X_test_imputed)


# Predict Probabilities
# We want the probability (score) that a GO term is present.
# predict_proba returns the probability for Class 0 (absent) and Class 1 (present).
print("7. Generating predictions...")
y_pred_probas = classifier.predict_proba(X_test_scaled)


# Format the Prediction Matrix
# Extract the probability for the positive class (index 1) from each of the 10 models
# and stack them back into a single matrix.
test_pred_matrix = np.column_stack([p[:, 1] for p in y_pred_probas])

# Create a DataFrame using the test proteins as the index and GO terms as columns
test_pred_df = pd.DataFrame(
    data=test_pred_matrix,
    index=aac_features_df.index,
    columns=Y_train.columns
)


# Convert to Long Format (Tidy Data)
# Transform the wide prediction table (224k x 10) into the long, tidy format (2.2M rows x 3 columns)
submission_ml_df = test_pred_df.melt(
    ignore_index=False,
    var_name='go_term',
    value_name='score'
).reset_index()


# Final Cleanup and Save
# Format score and save to TSV
submission_ml_df['score'] = submission_ml_df['score'].apply(lambda x: f"{x:.6f}")

timestamp = time.strftime("%Y%m%d-%H%M%S")
output_file = os.path.join(OUTPUT_DIR, f"submission_aac_lr_{timestamp}.tsv")

# Save the file (no header, no index)
submission_ml_df.to_csv(output_file, sep='\t', index=False, header=False)
submission_ml_df.to_csv("submission.tsv", sep='\t', index=False, header=False) # Overwrite generic name
print(f"8. Success! New submission file saved as: {output_file}")
print(f"   -> Total prediction rows: {len(submission_ml_df):,}")
```

```python
# Calculate Dipeptide Composition (DPC)
def calculate_dpc_features(fasta_path):
    """
    Reads the FASTA file and calculates the frequency of all 400 possible 
    dipeptide combinations (e.g., AA, AC, CA, YY).
    """
    print("9. Calculating Dipeptide Composition (DPC) features...")
    
    protein_features = {}
    amino_acids = 'ACDEFGHIKLMNPQRSTVWY' 
    dipeptides = [aa1 + aa2 for aa1 in amino_acids for aa2 in amino_acids] # 400 combinations
    
    for record in SeqIO.parse(fasta_path, "fasta"):
        parts = record.id.split('|')
        protein_id = parts[1] if len(parts) > 1 else record.id
        
        sequence = str(record.seq).upper().replace('*', '').replace('X', '')
        seq_length = len(sequence)
        
        # DPC requires at least two amino acids
        if seq_length < 2:
            continue
            
        # Count frequencies of pairs (overlapping pairs are used)
        counts = Counter([sequence[i:i+2] for i in range(seq_length - 1)])
        
        # Store the normalized frequencies
        features = {}
        # The normalization is (count / total number of pairs), which is (seq_length - 1)
        normalizer = seq_length - 1 
        
        for dp in dipeptides:
            features[dp] = counts[dp] / normalizer
            
        protein_features[protein_id] = features

    # Convert to DataFrame
    dpc_df = pd.DataFrame.from_dict(protein_features, orient='index')
    dpc_df.index.name = 'EntryID'
    
    print(f" -> DPC features calculated for {len(dpc_df):,} proteins.")
    return dpc_df
```

```python
# Execute DPC calculation
dpc_features_df = calculate_dpc_features(TEST_FASTA_PATH)

# Use the DPC features as the new X_train and retrain the model

# Align X (DPC) and Y (Labels)
X_train_merged_dpc = dpc_features_df.loc[Y_train_all.index] # Y_train_all is from before

# Preprocessing
# I need to re-fit the scaler and imputer since we have new features!
new_scaler = StandardScaler()
new_imputer = SimpleImputer(strategy='constant', fill_value=0)

X_train_imputed_dpc = new_imputer.fit_transform(X_train_merged_dpc)
X_train_scaled_dpc = new_scaler.fit_transform(X_train_imputed_dpc)

# Retrain the Classifier (Using the same fast LR model)
print("\n10. Training Multi-Label Logistic Regression with DPC...")
dpc_classifier = MultiOutputClassifier(LogisticRegression(
    solver='liblinear', random_state=42, max_iter=100
), n_jobs=-1)

start_time = time.time()
dpc_classifier.fit(X_train_scaled_dpc, Y_train_all) 
end_time = time.time()
print(f"Training completed in {end_time - start_time:.2f} seconds.")

# Store the new fitted scaler and imputer for use on the test set later
fitted_dpc_scaler = new_scaler
fitted_dpc_imputer = new_imputer
```

```python
# Predict and Generate DPC Submission File ---

# Test Features (X_test)
print("11. Preprocessing DPC test features...")
# Apply the SAME preprocessing steps fitted on the DPC training data.
X_test_imputed_dpc = fitted_dpc_imputer.transform(dpc_features_df)
X_test_scaled_dpc = fitted_dpc_scaler.transform(X_test_imputed_dpc)


# Predict Probabilities
print("12. Generating DPC predictions...")
# Predict using the dpc_classifier
y_pred_probas_dpc = dpc_classifier.predict_proba(X_test_scaled_dpc)


# Format the Prediction Matrix
# Extract the positive class probability (index 1) from each of the 10 models
test_pred_matrix_dpc = np.column_stack([p[:, 1] for p in y_pred_probas_dpc])

# Create a DataFrame for submission formatting
test_pred_df_dpc = pd.DataFrame(
    data=test_pred_matrix_dpc,
    index=dpc_features_df.index,  
    columns=Y_train_all.columns      
)


# Convert to Long Format (Tidy Data)
submission_dpc_df = test_pred_df_dpc.melt(
    ignore_index=False,
    var_name='go_term',
    value_name='score'
).reset_index()


# Cleanup and Save
submission_dpc_df['score'] = submission_dpc_df['score'].apply(lambda x: f"{x:.6f}")

timestamp = time.strftime("%Y%m%d-%H%M%S")
output_file_dpc = os.path.join(OUTPUT_DIR, f"submission_dpc_lr_{timestamp}.tsv")

# Save the file (no header, no index)
submission_dpc_df.to_csv(output_file_dpc, sep='\t', index=False, header=False)
submission_dpc_df.to_csv("submission.tsv", sep='\t', index=False, header=False) # Overwrite generic name
print(f"13. Success! New DPC submission file saved as: {output_file_dpc}")
print(f"   -> Total prediction rows: {len(submission_dpc_df):,}")
```
